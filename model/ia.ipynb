```python
import random
from joueur import Joueur

class IA(Joueur):
#ici IA hérite automatiquement du constructeur __init__() de Joueur, on a rien a modifier
    def lancer_de(self) -> int: #choisir 1 ou 2 des selon la Gare
            if self.ville.monuments[0].construit: #verifie si Gare est deja construite
                return random.choice([1, 2]) #1 ou 2 des
            else:
                return 1

    def choisir_joueur(self, partie: "Partie") -> "Joueur" | None: #choisir un adversaire au hasard
            adversaires = []
    
            for joueur in partie._joueurs:
                if joueur != self:
                    adversaires.append(joueur)
    
            if adversaires:
                return random.choice(adversaires)#ici l'IA choisit directement un objet Joueur au hasard sans saisir de numero
            else:
                return None
    
    def achat_cartes(self, partie: "Partie") -> str | None: #ici partie represente l'objet Partie transmis a l'IA 
        cartes_disponibles = []     #pour qu'elle puisse acceder a la reserve commune et consulter

        for nom in partie._reserve._cartes:
            carte = partie._reserve._cartes[nom]["objet"]

            if partie._reserve.est_disponible(nom) and self.argent >= carte.cout:
                cartes_disponibles.append(nom)

        if cartes_disponibles:
            return random.choice(cartes_disponibles + [None]) #selectionne un nom au hasard on ajoute a la liste None
        else:                                                 #cad choisir une carte a acheter ou decider de passer

            return None


    
    

```
